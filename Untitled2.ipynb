```python
def analyze_career_profile(user_skills_text):
    prompt = f"""
    You are a career advisor helping women professionals returning to work after a career break.

    Here is the person's background and skills:
    {user_skills_text}

    Please provide:
    1. Their key strengths (2-3 points)
    2. Skill gaps they should address for today's job market (2-3 points)
    3. Three relevant job roles they could target

    Keep the response clear, encouraging, and structured with headings.
    """

    response = client.models.generate_content(
        model="gemini-3-flash-preview",
        contents=prompt
    )
    return response.text

# Test karte hain
sample_input = "I have 5 years of experience in accounting and Excel. Took a 2-year career break for family. Comfortable with basic computer skills, want to get back into a data or finance related role."

result = analyze_career_profile(sample_input)
print(result)
```

```python
def analyze_career_profile(user_skills_text):
    prompt = f"""
    You are a career advisor helping women professionals returning to work after a career break.

    Here is the person's background and skills:
    {user_skills_text}

    Please provide:
    1. Their key strengths (2-3 points)
    2. Skill gaps they should address for today's job market (2-3 points)
    3. Three relevant job roles they could target

    Keep the response clear, encouraging, and structured with headings.
    """

    response = client.models.generate_content(
        model="gemini-3-flash-preview",
        contents=prompt
    )
    return response.text

# Test karte hain
sample_input = "I have 5 years of experience in accounting and Excel. Took a 2-year career break for family. Comfortable with basic computer skills, want to get back into a data or finance related role."

result = analyze_career_profile(sample_input)
print(result)
```

```python
import gradio as gr

def gradio_career_advisor(experience, career_gap, field, skills_background, desired_role):
    if not skills_background.strip():
        yield "⚠️ Kripya apni skills aur background zaroor likhein."
        return

    yield "⏳ **Aapki personalized career advice taiyar ho rahi hai... kripya kuch second wait karein.**"

    combined_input = f"""
    Years of Experience: {experience}
    Career Gap Duration: {career_gap}
    Field/Industry: {field}
    Skills and Background: {skills_background}
    Desired Role (if any): {desired_role if desired_role else 'Not specified'}
    """

    result = analyze_career_profile(combined_input)
    yield result

custom_theme = gr.themes.Soft(
    primary_hue="purple",
    secondary_hue="pink",
).set(
    body_background_fill="linear-gradient(135deg, #f5f3ff 0%, #fdf2f8 100%)"
)

with gr.Blocks(title="CareerBridge AI", theme=custom_theme) as demo:
    gr.Markdown(
        """
        # 🌉 CareerBridge AI
        ### Your personal guide back to the workforce, powered by Google Gemini
        Returning to work after a break can feel overwhelming. Tell us a bit about yourself,
        and get instant, personalized career guidance — your strengths, skill gaps, and best-fit roles.
        """
    )

    with gr.Row():
        experience = gr.Textbox(label="💼 Experience (years)", placeholder="Jaise: 5 years")
        career_gap = gr.Textbox(label="⏳ Career Gap Duration", placeholder="Jaise: 2 years")

    with gr.Row():
        field = gr.Textbox(label="🏢 Field / Industry", placeholder="Jaise: IT, Finance, Marketing")
        desired_role = gr.Textbox(label="🎯 Desired Role (optional)", placeholder="Jaise: Data Analyst")

    skills_background = gr.Textbox(
        label="📝 Apni skills aur background detail mein batao",
        placeholder="Certifications, key skills, previous responsibilities, achievements...",
        lines=5
    )

    submit_btn = gr.Button("✨ Get My Career Advice", variant="primary", size="lg")
    output = gr.Markdown(label="Career Advice")

    submit_btn.click(
        fn=gradio_career_advisor,
        inputs=[experience, career_gap, field, skills_background, desired_role],
        outputs=output
    )
demo.queue()
demo.launch(debug=True)
```
